### Задача 1. Найти корень числа

Алгоритм работает за O(log n), где n - число, корень из которого мы ищем, потому что мы на каждой итерации уменьшаем диапазон поиска в 2 раза, то есть на k-ой итерации длина не превосходит n/2^k, цикл продолжается в ходшем случае пока длина не стала меньше 1, то есть n/2^k<1 <=> 2^k>n <=> k>log_2(n). Каждая итерация выполняентся за O(1).

Правая граница поиска - это число, корень из которого необходимо найти, левая - 0, так как арифметический корень числа неотрицательный. Поиск продолжаем, пока не нашли точное число равное в квадрате target, или пока диапазон не пуст. На каждой итерации граница переопределяется в зависимости от значения квадрата среднего элемента: если он больше target, то сдвигаем правую границу, иначе - левую.

```python
def binarySearchSqrt(target):
    l = 0
    r = target
    while l<=r:
        middle = (l+r)//2
        if middle**2 > target:
            r = middle - 1
            continue
        if middle**2 < target:
            l = middle + 1
            continue
        return middle
    return r
```

```python
target = 49
print(f"target = {target}; sqrt(target) = {binarySearchSqrt(target)}")
target = 496
print(f"target = {target}; sqrt(target) = {binarySearchSqrt(target)}")
target = 17
print(f"target = {target}; sqrt(target) = {binarySearchSqrt(target)}")
```

### Задача 2. Очень лёгкая задача (два ксерокса)

Проверка на достаточность времени: mid//x + mid//y < n-1. Вычисляем сколько целых копий сделает первый и второй принтер при параллельной работе за mid минут и сравниваем с числом копий -1, так как первую копию делаем отдельно.

```python
def copyTime(n, x, y):
    l = 0
    r = (n - 1)*max(x, y)
    while l+1 < r:
        mid = (r + l)//2
        if mid//x + mid//y < n-1:
            l = mid
        else:
            r = mid
    return r + min(x, y)
```

```python
n = 4
x = 1
y = 2
print(f"N = {n}, x = {x}, y = {y}: result = {copyTime(n, x, y)}")
n = 10
x = 3
y = 3
print(f"N = {n}, x = {x}, y = {y}: result = {copyTime(n, x, y)}")
n = 5
x = 1
y = 10
print(f"N = {n}, x = {x}, y = {y}: result = {copyTime(n, x, y)}")
```

### Задача 3. Накормить животных

Выбранная стратегия даёт максимальное число накормленных животных, потому что мы сортируем массив животных и порций и сначал кормим животных, которые требуют меньше еды, меньшими порциями, тем самым мы не тратим большие порции на животных, которые едят мало. Реализован Жадный алгоритм, который и позволяет накормить максимальное число животных.

```python
def feedAnimals(animals, food):
    if len(animals) == 0 or len(food) == 0:
        return 0
    animals.sort()
    food.sort()
    count = 0
    for f in food:
        if f >= animals[count]:
            count+=1
        if count == len(animals):
            break
    return count
```

```python
animals = [2, 1, 3]
food = [1, 1, 1]
print(f"animals = {animals}, food = {food}, result = {feedAnimals(animals, food)}")
animals = [6, 2, 4]
food = [10, 2, 5]
print(f"animals = {animals}, food = {food}, result = {feedAnimals(animals, food)}")
animals = [1, 1, 1, 1, 1]
food = [100, 100]
print(f"animals = {animals}, food = {food}, result = {feedAnimals(animals, food)}")
```

### Задача 4. Найти разницу между двумя строками

По условию задачи строки a и b состоят из одинаковых символов, частоты всех кроме одного совпадают, у одного символа частота в строке b на один больше, чем в строке a. Про составлении хештаблицы для каждого симовла строки вычисляется его частота, затем в цикле по строке b происходит поиск символа в хештаблице и если он есть в таблице, то уменьшаем его частоту на один, если его нет в таблице - то это искомый символ, если после вычитания получили 0, то удаляем из таблицы. Тем самым из таблицы будут удалены все совпадающие символы.

```python
def extraLetter(a, b):
    hashMapA = dict()
    for i in range(len(a)):
        if a[i] in hashMapA:
            hashMapA[a[i]] += 1
        else:
            hashMapA[a[i]] = 1
    for i in range(len(b)):
        if b[i] in hashMapA:
            hashMapA[b[i]] -= 1
            if hashMapA[b[i]] == 0:
                del hashMapA[b[i]]
                continue
            continue
        return b[i]
    return ''
```

```python
a = "ansd"
b = "nafds"
print(f"a = {a}, b = {b}, result = {extraLetter(a, b)}")
a = ""
b = "n"
print(f"a = {a}, b = {b}, result = {extraLetter(a, b)}")
a = "dfgrbbas"
b = "dfrbasgbn"
print(f"a = {a}, b = {b}, result = {extraLetter(a, b)}")
a = "ddddd"
b = "dddddd"
print(f"a = {a}, b = {b}, result = {extraLetter(a, b)}")
```

### Задача 5. Сумма двух элементов массива

Мы запоминаем уже просмотренные элементы в хеш-таблице и за O(1) по ключу находим число, сумма с которым равна target. Если для текущего data[i] число diff встречалось ранее, то пара найдена. Это обусловлено в том числе симметричностью суммы. Такой подход позволяет избежать двойного цикла и достичь выполнения за O(n), в наивной реализации необходимо для каждого числа data[i] делать перебор всего массива и итоговое время O(n^2).

```python
def twoSum(data, target):
    cache = dict()
    for i in range(len(data)):
        cache[data[i]] = i
    for i in range(len(data)):
        diff = target - data[i]
        if diff in cache:
            if i != cache[diff]:
                return [i, cache[diff]]
    return []
```

```python
data = [2, 3, 5, 8, 2, 4, 5]
target = 13
print(f"data = {data}, target = {target}, result = {twoSum(data, target)}")
data = [3, 3, 3, 3]
target = 10
print(f"data = {data}, target = {target}, result = {twoSum(data, target)}")
data = [3, 3, 3, 3]
target = 6
print(f"data = {data}, target = {target}, result = {twoSum(data, target)}")
```

### Задача 6. Массив анаграмм

```python
def findAnagramm(strs):
    hashStrs = dict()
    for s in strs:
        sort = ''.join(sorted(s))
        if sort in hashStrs:
            hashStrs[sort].append(s)
        else:
            hashStrs[sort] = [s]
    result = []
    for key in hashStrs:
        result.append(hashStrs[key])
    return result
```

```python
strs = ["eat","tea","tan","ate","nat","bat"]
print(f"strs = {strs}, result = {findAnagramm(strs)}")
strs = ["a"]
print(f"strs = {strs}, result = {findAnagramm(strs)}")
strs = ["abc","bca","cab","acb","bac","cba"]
print(f"strs = {strs}, result = {findAnagramm(strs)}")
strs = ["abc","def","ghi"]
print(f"strs = {strs}, result = {findAnagramm(strs)}")
strs = ["", "", "a"]
print(f"strs = {strs}, result = {findAnagramm(strs)}")
strs = ["ab","abc","ba","cab","a"]
print(f"strs = {strs}, result = {findAnagramm(strs)}")
strs = ["aab","abb","ba","ab","aabb","bba"]
print(f"strs = {strs}, result = {findAnagramm(strs)}")
strs = ["abc","abc","abc","bca"]
print(f"strs = {strs}, result = {findAnagramm(strs)}")
strs = ["Abc","bca","ABC","cba","a b c"]
print(f"strs = {strs}, result = {findAnagramm(strs)}")
```

### Сортировка Шелла

```python
def shell_sort(arr):
    n = len(arr)
    gap = len(arr)//2
    while gap>0:
        for i in range(gap, n):
            m_gap = i
            while m_gap >= gap and arr[m_gap]<arr[m_gap - gap]:
                arr[m_gap], arr[m_gap - gap] = arr[m_gap - gap], arr[m_gap]
                m_gap -= gap
        gap = gap//2
    return arr
```

```python
data = [2, 3, 5, 8, 2, 4, 5]
print(f"data = {data}, result = {shell_sort(data)}")
data = [3, 4, 1, -5, 6, 2, 0, 9 , 25]
print(f"data = {data}, result = {shell_sort(data)}")
data = [3, 3, 3, 3]
print(f"data = {data}, result = {shell_sort(data)}")
```
